There are many ways to do this.  For example, you could use masking and loops.  However, you may find it easier and cleaner to use the `.groupby()` function, which is built-in to pandas dataframes and can be combined with other functions (documentation found [here](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html) and user guide found [here](https://pandas.pydata.org/pandas-docs/stable/user_guide/groupby.html#splitting-an-object-into-groups)). ````python
# Put your code here
import matplotlib.pyplot as plt

# Count tornadoes for each state and sort from most to least
state_counts = tornadoes.groupby("st").size().sort_values(ascending=False)

# Display the counts
print(state_counts)

# Create the bar plot
state_counts.plot(kind="bar", figsize=(14, 6))

plt.xlabel("State")
plt.ylabel("Number of Tornadoes")
plt.title("Total Number of Tornadoes by State")
plt.show()
#Source: Pandas DataFrame.groupby() documentation: Pandas documentation linked above
````

* Count the total number of tornadoes from each year across the entire US. Store these values in an appropriate container such as a pandas dataframe. (1 point)
* Using **masking**, filter the dataset to Michigan (MI) only and count the total number of tornadoes from each year in MI. (2 points)
* Make a plot (or plots) showing year on the x-axis and number of tornadoes on the y-axis, presenting **both** the US and Michigan trends. You may use two lines on the same axes or two subplots. ````python
# Put your code here
import matplotlib.pyplot as plt

# Count tornadoes per year across the entire US
us_yearly = tornadoes.groupby("yr").size()

# Mask the data to include only Michigan
mi_mask = tornadoes["st"] == "MI"
michigan = tornadoes[mi_mask]

# Count tornadoes per year in Michigan
mi_yearly = michigan.groupby("yr").size()

# Plot US tornadoes
plt.figure(figsize=(12, 5))
plt.plot(us_yearly.index, us_yearly.values)
plt.xlabel("Year")
plt.ylabel("Number of Tornadoes")
plt.title("Annual Number of Tornadoes in the United States")
plt.show()

# Plot Michigan tornadoes
plt.figure(figsize=(12, 5))
plt.plot(mi_yearly.index, mi_yearly.values)
plt.xlabel("Year")
plt.ylabel("Number of Tornadoes")
plt.title("Annual Number of Tornadoes in Michigan")
plt.show()
````
